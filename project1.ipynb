```python

# Loading our CSV file into a dataframe named 'df'
df = pd.read_csv("50_Startups.csv")

# Checking the dimensions (rows and columns) of our dataset
df.shape

# Counting how many missing (null) values are in each column
print("Missing values per column:")
print(df.isnull().sum())

# Counting how many completely duplicate rows exist in the dataset
print("\nTotal duplicate rows:")
print(df.duplicated().sum())


# Removing any duplicate rows from the dataset
df = df.drop_duplicates()

# Cleaning the 'State' column by removing extra spaces and making it lowercase
df['State'] = df['State'].str.strip().str.lower()

# Forcing 'Marketing Spend' to be numbers (errors turn into empty values)
df['Marketing Spend'] = pd.to_numeric(df['Marketing Spend'], errors='coerce')

# Printing a message to let us know the cleaning is done
print("\nCleaning completed successfully!")


# Drawing a simple random sample of exactly 20 rows
simple_sample = df.sample(n=20, random_state=7)

# Drawing a stratified sample taking 20% from each State group
stratified_sample = (
    df.groupby("State", group_keys=False)
    .apply(lambda g: g.sample(frac=0.20, random_state=7))
)



# Printing the average (mean) of the full population
print("\n Population Means :")
print("R&D Spend:", df["R&D Spend"].mean())
print("Marketing Spend:", df["Marketing Spend"].mean())

# Printing the average of the simple random sample
print("\n Simple Sample Means:")
print("R&D Spend:", simple_sample["R&D Spend"].mean())
print("Marketing Spend:", simple_sample["Marketing Spend"].mean())

# Printing  the average of the stratified sample
print("\n Stratified Sample Means:")
print("R&D Spend:", stratified_sample["R&D Spend"].mean())
print("Marketing Spend:", stratified_sample["Marketing Spend"].mean())
```

Because it guarantees that every single state gets its fair share in the sample. It stops important groups from being accidentally left out, which makes your final answers much more accurate than just picking names completely at random.
